```python
import pandas as pd
import geopandas as gpd
```

```python
# Panda settings for showing data (this is foremost done to more easily explore the data while processing it)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
```

```python
input_folder = "E://Proton_drive//EDUCATION//UNIGIS//Geodatabases//26_27//LRMB_v2024_monumental_trees_EN.csv"
output_folder = "E://Proton_drive//EDUCATION//UNIGIS//Geodatabases//26_27//"
```

Translating the inspection data into English

```python
monumental_trees = pd.read_csv(input_folder)
```

```python
monumental_trees.columns
```

```python
columns = ['OBJECTID', 'GlobalID', 'nr', 'locatie', 'boom_opm', 'naam_boom', 'plantjaar',
       'reden_opm', 'Laatste_waarneming', 'soort', 'sci', 'stamomtrek', 'hoogte',
       'status_label', 'x_rd', 'y_rd']

monumental_trees_sub = monumental_trees[columns]
```

```python
monumental_trees_sub['Laatste_waarneming'] = pd.to_datetime(monumental_trees_sub['Laatste_waarneming'], unit='ms')
```

```python
monumental_trees_sub.rename(columns={'locatie': 'location_NL'}, inplace=True)
monumental_trees_sub.rename(columns={'boom_opm': 'remarks_NL'}, inplace=True)
monumental_trees_sub.rename(columns={'naam_boom': 'name_tree_NL'}, inplace=True)

monumental_trees_sub.rename(columns={'reden_opm': 'remarks2_NL'}, inplace=True)
monumental_trees_sub.rename(columns={'Laatste_waarneming': 'last_observation'}, inplace=True)
monumental_trees_sub.rename(columns={'soort': 'species_NL'}, inplace=True)
monumental_trees_sub.rename(columns={'sci': 'scientific_name'}, inplace=True)
monumental_trees_sub.rename(columns={'stamomtrek': 'trunk_circumference'}, inplace=True)
monumental_trees_sub.rename(columns={'plantjaar': 'year_planted'}, inplace=True)
monumental_trees_sub.rename(columns={'hoogte': 'height'}, inplace=True)
monumental_trees_sub.rename(columns={'status_label': 'tree_status'}, inplace=True)
monumental_trees_sub.rename(columns={'x_rd': 'x_coordinate_rd'}, inplace=True)
monumental_trees_sub.rename(columns={'y_rd': 'y_coordinate_rd'}, inplace=True)
```

```python
monumental_trees_sub.head()
```

```python
print(monumental_trees_sub['year_planted'].unique().tolist())
monumental_trees_sub['year_planted'] = monumental_trees_sub['year_planted'].replace({
    'Leeftijd onbekend': 'Unknown',
    'vóór 1600': 'Before 1600'
})
```

```python
monumental_trees_sub['tree_status'] = monumental_trees_sub['tree_status'].replace({
    'Monumentaal': 'Monumental',
    'Dood/geveld': 'Dead/felled',
    'Afgemeld': 'Deregistered',
    'Potentieel monumentaal': 'Potentially monumental',
    'Afgekeurd': 'Rejected',
    'Aanmelding': 'Nominated'
})
```

```python
monumental_trees_sub.to_csv(output_folder+"LRMB_monumental_trees_EN.csv")
```
